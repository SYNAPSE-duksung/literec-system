# 교보문고 베스트셀러 리스트 중복 제거 및 중복 횟수 집계

같은 폴더(또는 지정 경로)에 있는 여러 개의 `교보문고_종합_베스트셀러_상품리스트*.xlsx` 파일을 모두 읽어
1. 하나로 합치고
2. `상품코드`(ISBN) 기준으로 중복을 제거한 상품 리스트를 만들고
3. 각 상품이 전체 파일들에서 몇 번 등장했는지 `중복횟수` 열로 추가하여
4. 새 엑셀 파일로 저장합니다.

```python
import glob
import pandas as pd

INPUT_DIR = "../data/raw/top-list"
FILE_PATTERN = "교보문고_종합_베스트셀러_상품리스트*.xlsx"

# 중복 판단 기준 컬럼 (상품코드=ISBN 기준으로 동일 상품 판단)
KEY_COLUMN = "상품코드"

OUTPUT_PATH = "../data/processed/교보문고_주간_베스트셀러_중복제거_리스트.xlsx"
```

```python
files = sorted(glob.glob(f"{INPUT_DIR}/{FILE_PATTERN}"))
print(f"발견된 파일 {len(files)}개")
for f in files:
    print(" -", f)
```

```python
# 모든 파일을 읽어 하나의 DataFrame으로 합치기 (어느 파일에서 왔는지 표시)
df_list = []
for f in files:
    tmp = pd.read_excel(f, dtype={KEY_COLUMN: str})
    tmp["출처파일"] = f
    df_list.append(tmp)

df_all = pd.concat(df_list, ignore_index=True)
print(f"합친 전체 행 수: {len(df_all)}")
df_all.head()
```

```python
# 상품코드 기준 중복 횟수 계산
dup_counts = df_all.groupby(KEY_COLUMN).size().rename("중복횟수")

# 중복 제거 (첫 번째 등장 행만 남김), 순위/출처파일처럼 파일마다 달라지는 정보는 참고용으로만 유지
df_unique = df_all.drop_duplicates(subset=KEY_COLUMN, keep="first").copy()

# 중복횟수 컬럼 붙이기
df_unique = df_unique.merge(dup_counts, on=KEY_COLUMN, how="left")

# 보기 좋게 정렬: 중복횟수 많은 순 -> 상품명 순
df_unique = df_unique.sort_values(by=["중복횟수", "상품명"], ascending=[False, True]).reset_index(drop=True)

print(f"중복 제거 후 고유 상품 수: {len(df_unique)}")
df_unique.head(20)
```

```python
# 중복된 상품(2회 이상 등장)만 확인
dup_only = df_unique[df_unique["중복횟수"] > 1]
print(f"여러 파일에 중복 등장한 상품 수: {len(dup_only)}")
dup_only[["상품코드", "상품명", "출판사", "중복횟수"]]
```

```python
# 결과 저장
df_unique.to_excel(OUTPUT_PATH, index=False)
print(f"저장 완료: {OUTPUT_PATH}")
```
